# Challenge 3: Robust Tagging Under Changing Detector Conditions

## What's the challenge?

A particle physics detector maps candidate particles in **η (eta) and φ (phi)** — pseudorapidity and azimuthal angle. When detector elements go offline — dead channels, noisy modules, alignment failures — particles whose trajectories pass through those regions simply don't show up in the event record.

Goal: **train a model whose anomaly detection stays robust as more of the detector drops out.**

At eval time the model sees events where affected candidates are missing (represented as zeros, same as padding). How you train for this is entirely up to you — augmentation strategy, architecture, objective. The only fixed constraints are the input format and how AUC is scored.

---

## Task

Design an anomaly detection model whose performance is robust as detector degradation occurs. Some ideas to improve robustness include:

- Giving the model degraded datasets during training
- Tracking current detector conditions to use as a conditional input to your model
- New training methods (such as contrastive learning with a degraded dataset to learn robustness)
- Using NN architectures with intrinsic / inherent robustness
- Your own ideas!

## How the baseline model works

The starting point (before you change anything) is a small Transformer encoder trained with a contrastive objective:

- Each event's PF candidates are embedded and passed through a Transformer (`TransformerEncoder` in `models.py`), producing one latent vector per event via a CLS token.
- A `Projector` head maps that latent into a space where a contrastive loss (`InfoNCELoss`, SupCon/InfoNCE-style) pulls same-class events closer together and pushes different-class events apart.
- A small linear classifier is trained alongside it (cross-entropy) on the four background classes (QCD, DY, TT, W+jets) — this just gives you a training-time accuracy/AUC signal to watch; the actual anomaly scoring happens later in `eval.py`, via a separate probe trained on the frozen latents to separate background from signal.
- Right now nothing in this pipeline is degradation-aware — `degradation.py` is a no-op stub, so the encoder only ever sees clean data during training. That's the gap your task is to close.

---

## Training data

- **Background only**: QCD, Drell-Yan, tt̄, W+jets — what the detector sees most of the time
- Each event: `[N, 7]` tensor of PF candidates with features `(pt, η, φ, dxy, dxy_sig, is_pf, pdgId)`. Padding rows (and missing candidates) have `pt == 0`.
- You're given the preprocessed tensors directly: `train/robust_tagging_train_data_small.pt`, `eval/robust_tagging_eval_small.pt`.

## Scoring

During evaluation the model will see both background and signal (anomaly) events and outputs a per-event anomaly score. In the eval script, evaluation is repeated multiple times for different severity of degradation applied to the evaluation dataset. The success of your model will be determined via **AUC vs. degradation severity** plots. A robust model maintains high AUC as the severity of detector errors increases. Your model's final score is the area under that curve.

## Section 3: Evaluate

`eval.py` loads a trained checkpoint, embeds the nominal eval set, trains a linear probe on those embeddings, then sweeps `--num_severities` degradation levels between `--sev_min` and `--sev_max` — applying **your own** `Degradation` (from `degradation.py`) at each fixed severity on the fly (no separate degraded-files step) — and re-embeds/re-scores at each one. Plots AUC vs. severity (the robustness curve) plus t-SNE diagnostics to `--outdir`.

This is a tool for testing your own approach locally, not the official scoring run. For judging, we'll degrade the eval set ourselves using a method we're not disclosing in advance — conceptually it kills off geometric regions of the η–φ plane (candidates landing in "dead" patches get dropped, mirroring how real detector failures cluster spatially rather than dropping candidates uniformly at random), but we're keeping the exact grid/probabilities to ourselves so nobody tunes specifically to it. Your model just needs to be robust to *that kind* of degradation in general.

Requires a checkpoint from Section 2 — set `latest_ckpt` above (or hardcode a path) before running this.

```python
assert latest_ckpt is not None, "train a model in Section 2 first (or set latest_ckpt manually)"

!python eval.py \
    --train_cfg {TRAIN_CFG} \
    --data_cfg {EVAL_DATA_CFG} \
    --encoder {latest_ckpt} \
    --data {EVAL_PT} \
    --outdir {EVAL_PLOTS_DIR} \
    --sev_min 0.0 \
    --sev_max 1.0 \
    --num_severities 10 \
    --grace_period 1000 \
    --diagnostics
```
